```python
# OTPW: load the 12-month dataset with spark.read.csv + gzip compression
df_otpw = (
    spark.read.format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load("dbfs:/mnt/mids-w261/OTPW_12M/OTPW_12M/OTPW_12M_2015.csv.gz")
)

display(df_otpw)
```

```python
from pyspark.sql import functions as F

total_rows = df_otpw.count()

null_summary = (
    df_otpw.select([
        F.count(F.when(F.col(c).isNull() | (F.col(c) == ""), c)).alias(c)
        for c in df_otpw.columns
    ])
    .toPandas()
    .T
    .reset_index()
    .rename(columns={'index': 'column', 0: 'null_count'})
)

null_summary["null_pct"] = (null_summary["null_count"] / total_rows * 100).round(2)
null_summary = null_summary.sort_values("null_pct", ascending=False)

display(null_summary)
```

```python
from pyspark.sql import functions as F

# delay rate by month
delay_by_month = (
    df_otpw.groupBy("MONTH")
    .agg(F.mean(F.col("DEP_DEL15").cast("double")).alias("delay_rate"))
    .orderBy("MONTH")
)
display(delay_by_month)

# delay rate by day of week
delay_by_day = (
    df_otpw.groupBy("DAY_OF_WEEK")
    .agg(F.mean(F.col("DEP_DEL15").cast("double")).alias("delay_rate"))
    .orderBy("DAY_OF_WEEK")
)
display(delay_by_day)
```

```python
delay_by_day_pd = delay_by_day.toPandas().sort_values("DAY_OF_WEEK")
import matplotlib.pyplot as plt
plt.figure(figsize=(8,5))
plt.bar(delay_by_day_pd["DAY_OF_WEEK"], delay_by_day_pd["delay_rate"], color="skyblue")
plt.title("Delay Rate by Day of Week (1=Mon, 7=Sun)")
plt.xlabel("Day of Week")
plt.ylabel("Delay Rate")
plt.show()
```

```python
df_otpw = df_otpw.withColumn("CRS_DEP_HOUR", (F.col("CRS_DEP_TIME").cast("int") / 100).cast("int"))

delay_by_hour = (
    df_otpw.groupBy("CRS_DEP_HOUR")
    .agg(F.mean(F.col("DEP_DEL15").cast("double")).alias("delay_rate"))
    .orderBy("CRS_DEP_HOUR")
)
display(delay_by_hour)
```

```python
import matplotlib.pyplot as plt

# Convert Spark result to pandas
delay_by_hour_pd = delay_by_hour.toPandas().sort_values("CRS_DEP_HOUR")

plt.figure(figsize=(10,5))
plt.plot(delay_by_hour_pd["CRS_DEP_HOUR"], delay_by_hour_pd["delay_rate"], marker="o", color="teal")
plt.title("Delay Rate by Scheduled Departure Hour")
plt.xlabel("Hour of Day (0–23)")
plt.ylabel("Delay Rate")
plt.grid(True)
plt.show()
```

```python
from pyspark.sql import functions as F

# Count number of flights per origin airport
airport_counts = (
    df_otpw.groupBy("ORIGIN")
    .count()
    .withColumn("pct", F.round(F.col("count") / df_otpw.count() * 100, 2))
    .orderBy(F.desc("count"))
)

display(airport_counts)
```

traffic concentrated among a few major airports and a long tail of low-volume regional airports.
This skew matters because it can affect model generalization and the patterns learned from ATL or ORD may not apply to airports with sparse data.

```python
import matplotlib.pyplot as plt

airport_counts_pd = airport_counts.toPandas()

plt.figure(figsize=(10,5))
plt.hist(airport_counts_pd["count"], bins=50, color="skyblue", edgecolor="black")
plt.title("Distribution of Flights per Origin Airport")
plt.xlabel("Number of Flights")
plt.ylabel("Number of Airports")
plt.show()
```

The dataset includes approximately 313 distinct origin airports. Flight volume is highly skewed: the top five hubs (ATL, ORD, DFW, DEN, LAX) account for roughly one-quarter of all flights, while more than half of airports contribute less than 0.1% each. We should consider this long tail during model training and avoid overfitting to high-traffic airports.

```python
from pyspark.sql import functions as F

# Calculate total flights and average delay rate for each origin airport
airport_summary = (
    df_otpw.groupBy("ORIGIN")
    .agg(
        F.count("*").alias("flight_count"),
        F.mean(F.col("DEP_DEL15").cast("double")).alias("delay_rate")
    )
    .orderBy(F.desc("flight_count"))
)

display(airport_summary)
```

```python
airport_summary_pd = airport_summary.toPandas()
```

```python
import matplotlib.pyplot as plt

# Create a new figure with specified size
plt.figure(figsize=(10,6))

# Scatter plot: x=number of flights per airport, y=average delay rate
plt.scatter(
    airport_summary_pd["flight_count"],
    airport_summary_pd["delay_rate"],
    alpha=0.6,
    color="teal"
)

# Set the plot title
plt.title("Airport Volume vs. Delay Rate")

# Label the x-axis
plt.xlabel("Number of Flights (per airport)")

# Label the y-axis
plt.ylabel("Average Delay Rate")

# Add grid lines for better readability
plt.grid(True)

# Annotate the 10 busiest airports with their codes
for i, row in airport_summary_pd.head(10).iterrows():
    plt.text(row["flight_count"], row["delay_rate"], row["ORIGIN"], fontsize=9, ha="left", va="bottom")

# Display the plot
plt.show()
```

```python
from pyspark.ml.feature import VectorAssembler
from pyspark.ml.stat import Correlation
from pyspark.sql import functions as F

# choose core numeric flight features (clean and relevant)
num_features = [
    "DEP_DELAY", "ARR_DELAY", "TAXI_OUT", "TAXI_IN",
    "AIR_TIME", "DISTANCE", "CRS_ELAPSED_TIME", "ACTUAL_ELAPSED_TIME"
]

# cast to double and drop nulls for correlation calculation
df_corr = df_otpw.select([F.col(c).cast("double").alias(c) for c in num_features]).na.drop()

# assemble vector
assembler = VectorAssembler(inputCols=num_features, outputCol="features")
df_vector = assembler.transform(df_corr)

# compute Pearson correlation matrix
corr_matrix = Correlation.corr(df_vector, "features", "pearson").collect()[0][0].toArray()

# display as a pandas dataframe for readability
import pandas as pd
pd.DataFrame(corr_matrix, index=num_features, columns=num_features)
```
